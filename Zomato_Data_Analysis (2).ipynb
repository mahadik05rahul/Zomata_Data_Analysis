```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_excel("zomato_restaurant_dataset.xlsx")
```

```python
print(df.head())
```

```python
city_count=df["City"].value_counts().head(10)
city_count.plot(kind='bar',figsize=(10,5))
plt.title("Top 10 Cities by Number of Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)
plt.show()

#City analysis
```

```python
rating_count=df["Rating"].value_counts().sort_index()
rating_count.plot(kind="bar",figsize=(10,6))
plt.title("Distribution of Restaurants Rating")
plt.xlabel("Rating")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

#Rating analysis
```

```python
df["Online Order"].value_counts().plot(kind="bar")
plt.title("Online Order Available")
plt.xlabel("Online Order")
plt.ylabel("Number of restaurants")
plt.show()
```

```python
Project Insights
1)City Analysis:-City which has the highest number of restaurants.
2)Rating Analysis:-Identify the most common value.
3)Online order:-Restaurants that accept online orders and not accept online orders.
```
